### 🔢 Problem

We want a neuron to learn to map:

```python
x = 2
y_true = 1
```

---

### 🧪 Code

```python
import math

# Sigmoid activation and its derivative
def sigmoid(z):
    return 1 / (1 + math.exp(-z))

def sigmoid_derivative(z):
    s = sigmoid(z)
    return s * (1 - s)

# Input and expected output
x = 2
y_true = 1

# Initialize weights and bias
w = 0.5
b = 0.0
learning_rate = 0.1

# Training loop
for epoch in range(20):
    # --- Forward pass ---
    z = w * x + b
    y_pred = sigmoid(z)
    loss = 0.5 * (y_pred - y_true) ** 2

    # --- Backward pass (manual gradients) ---
    dL_dy = y_pred - y_true
    dy_dz = sigmoid_derivative(z)
    dz_dw = x
    dz_db = 1

    # Chain rule
    dL_dw = dL_dy * dy_dz * dz_dw
    dL_db = dL_dy * dy_dz * dz_db

    # --- Update weights ---
    w -= learning_rate * dL_dw
    b -= learning_rate * dL_db

    print(f"Epoch {epoch+1}: Loss={loss:.4f}, w={w:.4f}, b={b:.4f}")
```

---

### 🧠 What It Shows

* Forward pass: `z = wx + b`, `y = sigmoid(z)`
* Loss: Squared error
* Backprop: Manual gradient calculation using the chain rule
* Weight update

## 📚 Theory Behind Backpropagation

---

### 🔧 1. **Goal** of Training

We want to find the weights $w$ and biases $b$ that minimize the **loss** — i.e., the difference between the predicted output $\hat{y}$ and the actual output $y$.

We do this by:

* **Computing gradients**: How much does the loss change with respect to each weight/bias?
* **Updating parameters**: Move weights in the direction that reduces loss.

---

### 📦 2. **Forward Pass**

In a simple neuron:

1. Compute the linear combination:

   $$
   z = wx + b
   $$

2. Apply the activation (sigmoid here):

   $$
   \hat{y} = \sigma(z) = \frac{1}{1 + e^{-z}}
   $$

3. Compute the loss:

   $$
   L = \frac{1}{2}(\hat{y} - y)^2
   $$

---

### 🔄 3. **Backward Pass (Backpropagation)**

We want to find how the **loss changes** with respect to the weight: $\frac{dL}{dw}$

We use the **chain rule**:

$$
\frac{dL}{dw} = \frac{dL}{d\hat{y}} \cdot \frac{d\hat{y}}{dz} \cdot \frac{dz}{dw}
$$

Break it down:

* $\frac{dL}{d\hat{y}} = \hat{y} - y$: how wrong we are
* $\frac{d\hat{y}}{dz} = \hat{y}(1 - \hat{y})$: derivative of sigmoid
* $\frac{dz}{dw} = x$: because $z = wx + b$

So:

$$
\frac{dL}{dw} = (\hat{y} - y) \cdot \hat{y}(1 - \hat{y}) \cdot x
$$

Same logic applies for bias:

$$
\frac{dL}{db} = (\hat{y} - y) \cdot \hat{y}(1 - \hat{y}) \cdot 1
$$

---

### 🔁 4. **Gradient Descent Update**

We update the parameters to reduce the loss:

$$
w = w - \eta \cdot \frac{dL}{dw}
$$

$$
b = b - \eta \cdot \frac{dL}{db}
$$

Where $\eta$ is the **learning rate** — how big a step we take.

---

### 🧠 Big Picture

* Forward pass: compute prediction and loss
* Backward pass: compute how to change weights to reduce loss
* Update: apply changes (gradient descent)
* Repeat until the loss is small

---




**Backpropagation** (short for "backward propagation of errors") is a fundamental algorithm used to train artificial neural networks in deep learning. It computes the **gradient of the loss function** with respect to each weight in the network, enabling the use of optimization techniques like **gradient descent** to minimize the loss.

---

## 🔁 Key Steps of Backpropagation

Assume we have a neural network with layers:

1. **Forward Pass**:

   * Compute the output (prediction) of the network by passing the input data forward through the layers.
   * Calculate the **loss** (error) between predicted and actual output.

2. **Backward Pass**:

   * Use the **chain rule** of calculus to compute gradients of the loss function with respect to each parameter (weights and biases).
   * Gradients are computed layer by layer, moving **from the output layer back to the input layer**.

3. **Weight Update**:

   * Use the gradients to update the weights:

     $$
     w := w - \eta \cdot \frac{\partial L}{\partial w}
     $$

     where $\eta$ is the learning rate and $\frac{\partial L}{\partial w}$ is the gradient of the loss with respect to the weight.

---

## 🧠 Mathematical Intuition

For a network layer:

* Input: $x$
* Weights: $W$
* Bias: $b$
* Activation: $a = f(Wx + b)$

During backpropagation:

1. Compute the derivative of the loss $L$ with respect to the output.
2. Use the chain rule to propagate this derivative back to compute $\frac{\partial L}{\partial W}$, $\frac{\partial L}{\partial b}$, and $\frac{\partial L}{\partial x}$.

---

## 🔄 Example (1 Hidden Layer Network)

Given:

* Input $x$
* Hidden layer weights $W_1$, activation function $f$
* Output layer weights $W_2$, output $y$
* Loss function: $L(y, \hat{y})$

Backpropagation involves computing:

* $\frac{\partial L}{\partial W_2}$
* $\frac{\partial L}{\partial W_1}$
* Then updating weights using these gradients.
