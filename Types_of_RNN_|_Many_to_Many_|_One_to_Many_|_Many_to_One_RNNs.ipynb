## ✅ Four Types of RNN Architectures

| Type | Input | Output | Use Case |
|------|-------|--------|---------|
| **1. One-to-One** | Single input | Single output | Standard feedforward neural network |
| **2. One-to-Many** | Single input | Sequence of outputs | e.g., Image captioning (image → sentence) |
| **3. Many-to-One** | Sequence of inputs | Single output | e.g., Sentiment analysis (sentence → label) |
| **4. Many-to-Many** | Sequence of inputs | Sequence of outputs | e.g., Machine translation (sentence → sentence) |

> 🔁 Note: "Many-to-Many" can be further split into:
> - **Synchronous**: Each input corresponds to one output (e.g., speech recognition)
> - **Asynchronous**: Input and output sequences have different lengths (e.g., translation)

### ❓Q2: Explain Many-to-One RNN with an example.

**✅ Candidate Answer:**

- **Input**: A sequence of vectors (e.g., words in a sentence)  
- **Output**: A single vector (e.g., class label)

🔁 The RNN processes each word step by step, and the final hidden state is used to make a prediction.

🧠 **Example**:  
**Sentiment Analysis**  
Input: `"I love this movie"` → `[x₁, x₂, x₃, x₄]`  
Output: `Positive` (single label)

🧩 Architecture:
- Use `LSTM()` or `GRU()` without `return_sequences`
- Final hidden state → passed to Dense layer for classification

### ❓Q3: What is a One-to-Many RNN? Give a real-world use case.

**✅ Candidate Answer:**

- **Input**: A single vector (e.g., image embedding or start token)  
- **Output**: A sequence of vectors (e.g., words in a caption)

🔁 At each step, the RNN generates one output and feeds it back as input for the next step.

🧠 **Example**:  
**Image Captioning**  
Input: Image (encoded as a vector)  
Output: `"A dog playing in the park"`

🔁 Or **Music Generation**  
Input: Initial note  
Output: Melody sequence

🧩 Uses **auto-regressive decoding**: output at time *t* becomes input at *t+1*

### ❓Q4: Describe Many-to-Many RNN. Are there subtypes?

**✅ Candidate Answer:**

Yes — **Many-to-Many** means:
- **Input**: Sequence
- **Output**: Sequence

But there are **two key subtypes**:

#### 1. **Synchronous (Equal Length)**
- Input and output have same length
- One output per input step

🧠 Example:  
**Part-of-Speech Tagging**  
Input: `["The", "cat", "runs"]`  
Output: `["DET", "NOUN", "VERB"]`

🧩 Model: `LSTM(return_sequences=True)` → output at every step

#### 2. **Asynchronous (Variable Length)**
- Input and output sequences can differ in length

🧠 Example:  
**Machine Translation**  
Input: `"Je t’aime"` (3 words)  
Output: `"I love you"` (3 words — but could be different)

🧩 Uses **Encoder-Decoder architecture**:
- Encoder: processes input → final state
- Decoder: uses that state to generate output sequence

### ❓Q5: What is a One-to-One RNN? Is it really an RNN?

**✅ Candidate Answer:**

- **Input**: One value  
- **Output**: One value  

But — and this is important — **One-to-One is not truly an RNN**.

It's just a **standard feedforward network** with no sequence processing.

🔁 True RNNs must have **recurrence over time steps** — meaning they process sequences and maintain a hidden state across steps.

👉 So while "One-to-One" appears in diagrams for completeness, it's **not a real RNN** unless part of a larger sequential system.

### ❓Q7: When would you choose one RNN type over another?

**✅ Candidate Answer:**

Choose based on your **task’s input-output structure**:

| Task | RNN Type |
|------|---------|
| Sentiment Analysis | **Many-to-One** |
| Text Classification | **Many-to-One** |
| Image/Video Captioning | **One-to-Many** |
| Music/Poem Generation | **One-to-Many** |
| POS Tagging, NER | **Many-to-Many (Synchronous)** |
| Machine Translation, Summarization | **Many-to-Many (Asynchronous)** |

👉 Never use One-to-One for sequential data — it has no memory.

### ❓Q8: Can a single model combine multiple RNN types?

**✅ Candidate Answer:**

Yes — especially in **encoder-decoder** models.

For example, in **machine translation**:
- **Encoder**: Many-to-One (processes input sentence → context vector)
- **Decoder**: One-to-Many (uses context to generate output sequence)

So the full system is **Many-to-Many**, but internally combines:
- Many-to-One (encoder)
- One-to-Many (decoder)

This hybrid approach is powerful and widely used.
