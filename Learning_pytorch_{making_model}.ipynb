##Part 1 – Manual everything (no nn.Module, no optimizers)


```python
import torch

# Create toy dataset
torch.manual_seed(0)
X = torch.rand(100, 1)
y = 3 * X + 2 + 0.1 * torch.randn(100, 1)  # Noise added

# Initialize parameters manually
W = torch.randn(1, requires_grad=True)
b = torch.randn(1, requires_grad=True)

# Hyperparameters
lr = 0.1
epochs = 200

for epoch in range(epochs):
    # Forward pass (no nn.Linear)
    y_pred = W * X + b  # Linear model

    # Loss (MSE) without nn.MSELoss
    loss = ((y_pred - y) ** 2).mean()

    # Backward pass
    loss.backward()

    # Gradient descent update manually
    with torch.no_grad():
        W -= lr * W.grad
        b -= lr * b.grad

    # Zero gradients manually
    W.grad.zero_()
    b.grad.zero_()

    if (epoch + 1) % 10 == 0:
        print(f"Epoch {epoch+1}, Loss: {loss.item():.4f}, W: {W.item():.4f}, b: {b.item():.4f}")

print("\nManual model learned: W =", W.item(), "b =", b.item())
```

##Part 2 – Using torch.nn & torch.optim


```python
import torch
import torch.nn as nn
import torch.optim as optim

# Same dataset
torch.manual_seed(0)
X = torch.rand(100, 1)
y = 3 * X + 2 + 0.1 * torch.randn(100, 1)

# Define model using nn.Module
class LinearModel(nn.Module):
    def __init__(self):
        super(LinearModel, self).__init__()
        self.linear = nn.Linear(1, 1, bias=True)  # One input, one output

    def forward(self, x):
        return self.linear(x)

model = LinearModel()

# Loss & Optimizer
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.1)

# Training loop
epochs = 200
for epoch in range(epochs):
    y_pred = model(X)
    loss = criterion(y_pred, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if (epoch + 1) % 10 == 0:
        # Extract parameters for printing
        W, b = model.linear.weight.item(), model.linear.bias.item()
        print(f"Epoch {epoch+1}, Loss: {loss.item():.4f}, W: {W:.4f}, b: {b:.4f}")

print("\nNN module model learned:", list(model.parameters()))
```

```python
# create model class
import torch
import torch.nn as nn

class Model(nn.Module):

  def __init__(self, num_features):

    super().__init__()
    self.network = nn.Sequential(
        nn.Linear(num_features, 3),
        nn.ReLU(),
        nn.Linear(3, 1),
        nn.Sigmoid()
    )

  def forward(self, features):

    out = self.network(features)

    return out
```

```python
# create dataset
features = torch.rand(10,5)

# create model
model = Model(features.shape[1])

# call model for forward pass
# model.forward(features)
model(features)
```
