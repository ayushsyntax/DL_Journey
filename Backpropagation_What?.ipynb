
## **Step-by-Step Mathematics of Backpropagation**

Suppose we have a neural network with a single hidden layer. For simplicity:

- Let input vector: $$ \mathbf{x} $$
- Weights (input to hidden): $$ \mathbf{W}_1 $$, biases: $$ \mathbf{b}_1 $$
- Hidden activations: $$ \mathbf{h} = f(\mathbf{W}_1 \mathbf{x} + \mathbf{b}_1) $$ where $$ f $$ is the activation function (e.g., sigmoid, ReLU)
- Weights (hidden to output): $$ \mathbf{W}_2 $$, biases: $$ \mathbf{b}_2 $$
- Output: $$ \hat{y} = g(\mathbf{W}_2 \mathbf{h} + \mathbf{b}_2) $$ (where $$ g $$ may be another activation, often identity or softmax)
- Ground truth: $$ y $$
- Loss function: $$ L(\hat{y}, y) $$, e.g. mean squared error: $$ L = \frac{1}{2}(\hat{y} - y)^2 $$

### **1. Forward Pass**
1. Compute $$ \mathbf{h} = f(\mathbf{W}_1 \mathbf{x} + \mathbf{b}_1) $$
2. Compute $$ \hat{y} = g(\mathbf{W}_2 \mathbf{h} + \mathbf{b}_2) $$
3. Compute loss: $$ L = \frac{1}{2}(\hat{y} - y)^2 $$

### **2. Backward Pass (Gradient Computation)**

**Goal:** Find how much to update each weight

#### a. Output Layer Gradients

- Compute gradient of loss w.r.t output:  
  $$ \frac{\partial L}{\partial \hat{y}} = \hat{y} - y $$
- Gradient w.r.t $$ \mathbf{W}_2 $$ (using chain rule):  
  $$ \frac{\partial L}{\partial \mathbf{W}_2} = \frac{\partial L}{\partial \hat{y}} \cdot \frac{\partial \hat{y}}{\partial \mathbf{W}_2} = (\hat{y} - y) \cdot \mathbf{h}^T $$
- Gradient w.r.t $$ \mathbf{b}_2 $$:  
  $$ \frac{\partial L}{\partial \mathbf{b}_2} = \hat{y} - y $$

#### b. Hidden Layer Gradients

- Compute gradient from output layer, backward through activation and weights:
  $$ \frac{\partial L}{\partial \mathbf{h}} = (\hat{y} - y) \cdot \mathbf{W}_2^T $$
- Multiply by derivative of activation function (element-wise):  
  If $$ f $$ is sigmoid, its derivative is $$ f'(\mathbf{z}) = f(\mathbf{z}) \cdot (1 - f(\mathbf{z})) $$  
  $$ \frac{\partial L}{\partial (\mathbf{W}_1 \mathbf{x} + \mathbf{b}_1)} = \frac{\partial L}{\partial \mathbf{h}} \odot f'(\mathbf{W}_1 \mathbf{x} + \mathbf{b}_1) $$
- Now, gradient w.r.t $$ \mathbf{W}_1 $$:  
  $$ \frac{\partial L}{\partial \mathbf{W}_1} = \left[ \frac{\partial L}{\partial (\mathbf{W}_1 \mathbf{x} + \mathbf{b}_1)} \right] \cdot \mathbf{x}^T $$
- Gradient w.r.t $$ \mathbf{b}_1 $$:  
  $$ \frac{\partial L}{\partial \mathbf{b}_1} = \frac{\partial L}{\partial (\mathbf{W}_1 \mathbf{x} + \mathbf{b}_1)} $$

### **3. Weight Updates**

Using gradient descent, update each parameter:
$$
\text{new parameter} = \text{old parameter} - \eta \times \text{gradient}
$$
Where $$ \eta $$ is the learning rate.

For example:
- $$ \mathbf{W}_2 \leftarrow \mathbf{W}_2 - \eta \cdot \frac{\partial L}{\partial \mathbf{W}_2} $$
- $$ \mathbf{b}_2 \leftarrow \mathbf{b}_2 - \eta \cdot \frac{\partial L}{\partial \mathbf{b}_2} $$
- $$ \mathbf{W}_1 \leftarrow \mathbf{W}_1 - \eta \cdot \frac{\partial L}{\partial \mathbf{W}_1} $$
- $$ \mathbf{b}_1 \leftarrow \mathbf{b}_1 - \eta \cdot \frac{\partial L}{\partial \mathbf{b}_1} $$

**Summary:**  
- **Forward pass:** Compute predictions and loss.
- **Backward pass:** Use chain rule to find the loss’s gradient w.r.t each parameter.
- **Update:** Nudge each parameter in the direction that reduces the loss, using the calculated gradients.  
Repeat for each batch or sample!

That’s backpropagation—*efficient error-driven learning via calculus and the chain rule*!
