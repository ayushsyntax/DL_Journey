### 🔹 Step-by-Step Forward Pass

1. **Input: X**
   - The input is typically a 2D image (e.g., 6×6 as in previous slide).
   - Represented as `X`.

2. **Convolution Operation: ⊗**
   - Applies a convolution filter (kernel) `w₁` to `X`, plus bias `b₁`.
   - Output: `z₁ = X * w₁ + b₁`
     - `*` denotes convolution.
     - This produces a feature map (pre-activation).

3. **ReLU Activation: z₁ → A₁**
   - Element-wise ReLU: `A₁ = relu(z₁) = max(0, z₁)`
   - Introduces non-linearity.

4. **Max Pooling: A₁ → P₁**
   - Downsamples the feature map using max pooling (e.g., 2×2 window, stride=2).
   - Reduces spatial dimensions while retaining important features.

5. **Flatten: P₁ → F**
   - Converts the 2D pooled feature map into a 1D vector.
   - E.g., if `P₁` is 2×2 → `F` is a vector of size 4.

6. **Fully Connected Layer: F → z₂**
   - Linear transformation: `z₂ = F ⋅ w₂ + b₂`
   - `w₂` is the weight matrix connecting flattened features to output neuron.
   - `b₂` is the bias.

7. **Sigmoid Activation: z₂ → A₂**
   - `A₂ = sigmoid(z₂) = 1 / (1 + exp(-z₂))`
   - Outputs a probability between 0 and 1.
   - Used for binary classification.

8. **Loss: A₂ → L**
   - Final output `A₂` is compared to true label `y` to compute loss.
   - Loss function: Binary Cross-Entropy:
     $$
     L = -y \log(A_2) - (1-y)\log(1-A_2)
     $$

9. **Target: A₂ ≈ y**
   - Goal is to make predicted probability `A₂` close to true label `y`.
   - Shown as: `A₂ ≈ y` (written at top right)

## ✅ Step 1: Forward Pass (Recap)

From previous slides:

$$
\begin{align*}
z_1 &= \text{Conv}(X, w_1) + b_1 \\
A_1 &= \text{ReLU}(z_1) \\
P_1 &= \text{maxpool}(A_1) \\
F &= \text{flatten}(P_1) \\
z_2 &= w_2 F + b_2 \\
A_2 &= \sigma(z_2) \\
L &= -y \log(A_2) - (1-y)\log(1-A_2)
\end{align*}
$$

---

## 🔄 Step 2: Backward Pass – Chain Rule

Backpropagation uses the **chain rule** to compute gradients from output to input.

We start from the **loss `L`** and move backward through layers.

---

### 🔹 1. Gradient w.r.t. $ w_2 $ and $ b_2 $

#### Compute $ \frac{\partial L}{\partial w_2} $:

$$
\frac{\partial L}{\partial w_2} = \frac{\partial L}{\partial A_2} \times \frac{\partial A_2}{\partial z_2} \times \frac{\partial z_2}{\partial w_2}
$$

Now substitute:

- $ \frac{\partial L}{\partial A_2} = \frac{A_2 - y}{A_2(1-A_2)} $ ← derivative of binary cross-entropy
- $ \frac{\partial A_2}{\partial z_2} = \sigma'(z_2) = A_2(1 - A_2) $
- $ \frac{\partial z_2}{\partial w_2} = F $ ← because $ z_2 = w_2 F + b_2 $

So:
$$
\frac{\partial L}{\partial w_2} = (A_2 - y) \times F
$$

> This is the **gradient of loss w.r.t. the FC layer weights**.

#### Compute $ \frac{\partial L}{\partial b_2} $:

$$
\frac{\partial L}{\partial b_2} = \frac{\partial L}{\partial A_2} \times \frac{\partial A_2}{\partial z_2} \times \frac{\partial z_2}{\partial b_2}
= (A_2 - y) \times A_2(1 - A_2) \times 1 = A_2 - y
$$

> So: $ \frac{\partial L}{\partial b_2} = A_2 - y $

---

### 🔹 2. Gradient w.r.t. $ w_1 $ and $ b_1 $

Now we go **back further**: from $ w_2 $ → $ F $ → $ P_1 $ → $ A_1 $ → $ z_1 $ → $ w_1 $

We need:
$$
\frac{\partial L}{\partial w_1} = \frac{\partial L}{\partial A_2} \times \frac{\partial A_2}{\partial z_2} \times \frac{\partial z_2}{\partial F} \times \frac{\partial F}{\partial P_1} \times \frac{\partial P_1}{\partial A_1} \times \frac{\partial A_1}{\partial z_1} \times \frac{\partial z_1}{\partial w_1}
$$

Let’s compute each term:

| Term | Value |
|------|-------|
| $ \frac{\partial L}{\partial A_2} $ | $ A_2 - y $ |
| $ \frac{\partial A_2}{\partial z_2} $ | $ A_2(1 - A_2) $ |
| $ \frac{\partial z_2}{\partial F} $ | $ w_2 $ (transpose if needed) |
| $ \frac{\partial F}{\partial P_1} $ | Identity (flattening is linear) |
| $ \frac{\partial P_1}{\partial A_1} $ | **Gradient of max pooling** — only non-zero where max occurred |
| $ \frac{\partial A_1}{\partial z_1} $ | $ \text{ReLU}'(z_1) = 1 $ if $ z_1 > 0 $, else 0 |
| $ \frac{\partial z_1}{\partial w_1} $ | $ X $ (input image), but convolved in reverse |

---

### 🔎 Key Insight: Backprop Through Layers

The diagram shows that the gradient flows backward as:

> **Loss → Sigmoid → FC → Flatten → Max Pool → ReLU → Conv**

And the chain rule combines all these partial derivatives.

Let’s write the full expression again:

$$
\frac{\partial L}{\partial w_1} = \underbrace{(A_2 - y)}_{\partial L / \partial A_2} \times \underbrace{A_2(1 - A_2)}_{\partial A_2 / \partial z_2} \times \underbrace{w_2}_{\partial z_2 / \partial F} \times \underbrace{1}_{\partial F / \partial P_1} \times \underbrace{\text{grad\_pool}}_{\partial P_1 / \partial A_1} \times \underbrace{\text{ReLU}'(z_1)}_{\partial A_1 / \partial z_1} \times \underbrace{X}_{\partial z_1 / \partial w_1}
$$

But note:
- $ \frac{\partial z_1}{\partial w_1} $ is not just `X`, but involves **convolution with input**.
- In practice, this is computed using **cross-correlation** of the gradient with the input.

---

### 🧠 How It Works Practically

#### For $ \frac{\partial L}{\partial w_1} $:
- Start with gradient at output: $ \delta = A_2 - y $
- Propagate back through sigmoid: $ \delta_z = \delta \cdot A_2(1 - A_2) $
- Multiply by $ w_2 $: gives gradient w.r.t. `F`
- Unflatten → get gradient w.r.t. `P₁`
- Back through max pool: **only pass gradient to max element**
- Back through ReLU: multiply by `1` or `0` depending on whether $ z_1 > 0 $
- Finally, convolve this gradient with the **original input `X`** to get $ \frac{\partial L}{\partial w_1} $

This is called **convolutional backpropagation**.

---

### ✅ Summary of Gradients

| Parameter | Gradient Formula |
|---------|------------------|
| $ \frac{\partial L}{\partial w_2} $ | $ (A_2 - y) \cdot F $ |
| $ \frac{\partial L}{\partial b_2} $ | $ A_2 - y $ |
| $ \frac{\partial L}{\partial w_1} $ | Convolution of input `X` with gradient from previous layers |
| $ \frac{\partial L}{\partial b_1} $ | Sum of gradients over spatial dimensions (after backprop through all layers) |

---

### 📌 Final Answer: **How?**

> **Backpropagation computes gradients using the chain rule**, moving backward through each layer:
>
> 1. Compute gradient of loss w.r.t. final output ($ \partial L / \partial A_2 $)
> 2. Use chain rule to propagate gradients backward:
>    - Through sigmoid → FC → flatten → max pool → ReLU → convolution
> 3. At each layer, apply the appropriate derivative:
>    - Sigmoid: $ \sigma'(z) = \sigma(z)(1-\sigma(z)) $
>    - ReLU: $ \mathbb{1}(z > 0) $
>    - Max pool: gradient goes only to max element
>    - Conv: use correlation with input
> 4. Update parameters: $ w \leftarrow w - \eta \frac{\partial L}{\partial w} $

## ✅ 1. Gradient w.r.t. $ w_2 $

$$
\boxed{ \frac{\partial L}{\partial w_2} = (A_2 - y) F^T }
$$

### 🔎 Explanation:
- Recall:  
  $$
  z_2 = w_2 F + b_2 \\
  A_2 = \sigma(z_2)
  $$

- Loss: Binary cross-entropy → derivative w.r.t. $ A_2 $ is $ A_2 - y $

- Chain rule:
  $$
  \frac{\partial L}{\partial w_2} = \frac{\partial L}{\partial A_2} \cdot \frac{\partial A_2}{\partial z_2} \cdot \frac{\partial z_2}{\partial w_2}
  $$

  But:
  - $ \frac{\partial L}{\partial A_2} = A_2 - y $
  - $ \frac{\partial A_2}{\partial z_2} = A_2(1 - A_2) $
  - $ \frac{\partial z_2}{\partial w_2} = F $ (since $ z_2 = w_2 F + b_2 $)

So:
$$
\frac{\partial L}{\partial w_2} = (A_2 - y) \cdot A_2(1 - A_2) \cdot F
$$

But wait — we have an extra term!

However, in **practice**, when using **sigmoid + binary cross-entropy**, the combined gradient simplifies to:

$$
\frac{\partial L}{\partial z_2} = A_2 - y
$$

Because:
$$
\frac{\partial L}{\partial z_2} = \frac{\partial L}{\partial A_2} \cdot \frac{\partial A_2}{\partial z_2} = (A_2 - y) \cdot A_2(1 - A_2) \cdot \text{(but actually...)}
$$

Wait — let's double-check:

Actually, the **derivative of binary cross-entropy with sigmoid output** is:

$$
\frac{\partial L}{\partial z_2} = \sigma(z_2) - y = A_2 - y
$$

✅ So yes: $ \frac{\partial L}{\partial z_2} = A_2 - y $

Then:
$$
\frac{\partial L}{\partial w_2} = \frac{\partial L}{\partial z_2} \cdot \frac{\partial z_2}{\partial w_2} = (A_2 - y) \cdot F^T
$$

> Why $ F^T $? Because:
> - $ w_2 $ has shape $ (1, 4) $
> - $ F $ has shape $ (4, 1) $
> - $ z_2 = w_2 F + b_2 $
> - So $ \frac{\partial z_2}{\partial w_2} = F^T $ → shape $ (1, 4) $

Thus:
$$
\frac{\partial L}{\partial w_2} = (A_2 - y) \cdot F^T
$$

→ Shape: $ (1, 4) $

---

## ✅ 2. Gradient w.r.t. $ b_2 $

$$
\boxed{ \frac{\partial L}{\partial b_2} = A_2 - y }
$$

### 🔎 Explanation:
- $ z_2 = w_2 F + b_2 $
- $ \frac{\partial z_2}{\partial b_2} = 1 $
- So:
  $$
  \frac{\partial L}{\partial b_2} = \frac{\partial L}{\partial z_2} \cdot \frac{\partial z_2}{\partial b_2} = (A_2 - y) \cdot 1 = A_2 - y
  $$

→ Scalar value (shape: $ (1,) $)

Below the gradients, we see:

$$
\frac{\partial L}{\partial w_2} = (A_2 - y) \cdot F^T
$$

- $ (1,m) \times (m,4) = (1,4) $

✅ Same as before — but now averaged over batch.

This is how **mini-batch gradient descent** works.

### Forward Pass:

- $ \frac{\partial L}{\partial w_1} $

---

## ✅ Step 1: Loss Derivative

First, compute $ \frac{\partial L}{\partial A_2} $

For binary cross-entropy:
$$
L = -y \log(A_2) - (1-y)\log(1-A_2)
$$

Take derivative w.r.t. $ A_2 $:
$$
\frac{\partial L}{\partial A_2} = -\frac{y}{A_2} + \frac{1-y}{1-A_2}
$$

But simplify:
$$
\frac{\partial L}{\partial A_2} = \frac{A_2 - y}{A_2(1 - A_2)}
$$

Now, since $ A_2 = \sigma(z_2) $, and $ \sigma'(z_2) = A_2(1 - A_2) $, then:

$$
\frac{\partial L}{\partial z_2} = \frac{\partial L}{\partial A_2} \cdot \frac{\partial A_2}{\partial z_2} = \frac{A_2 - y}{A_2(1 - A_2)} \cdot A_2(1 - A_2) = A_2 - y
$$

✅ So:  
$$
\boxed{ \frac{\partial L}{\partial z_2} = A_2 - y }
$$

This is the **key simplification** — the gradient of loss w.r.t. pre-activation becomes just $ A_2 - y $.

---

## ✅ Step 2: Gradient w.r.t. $ w_2 $ and $ b_2 $

We already know:
- $ z_2 = w_2 F + b_2 $
- $ \frac{\partial z_2}{\partial w_2} = F^T $ → shape $ (4,1) $ if $ F $ is $ (4,1) $
- $ \frac{\partial z_2}{\partial b_2} = 1 $

So:
$$
\frac{\partial L}{\partial w_2} = \frac{\partial L}{\partial z_2} \cdot \frac{\partial z_2}{\partial w_2} = (A_2 - y) \cdot F^T
$$
$$
\frac{\partial L}{\partial b_2} = (A_2 - y) \cdot 1 = A_2 - y
$$

These were covered in previous slides.

---

## ✅ Step 3: Backprop Through Flatten Layer

Now go backward from $ z_2 $ → $ F $

We need:
$$
\frac{\partial L}{\partial F} = \frac{\partial L}{\partial z_2} \cdot \frac{\partial z_2}{\partial F} = (A_2 - y) \cdot w_2
$$

Because $ z_2 = w_2 F + b_2 $, so $ \frac{\partial z_2}{\partial F} = w_2 $

→ So:
$$
\boxed{ \frac{\partial L}{\partial F} = (A_2 - y) \cdot w_2 }
$$

Shape: $ (1,4) $ if $ w_2 $ is $ (1,4) $

---

## ✅ Step 4: Backprop Through Max Pooling

Now: $ F = \text{flatten}(P_1) $

So $ \frac{\partial L}{\partial P_1} = \text{reshape}\left( \frac{\partial L}{\partial F} \right) $

But more importantly:
$$
\frac{\partial L}{\partial P_1} = \frac{\partial L}{\partial F} \cdot \frac{\partial F}{\partial P_1}
$$

Since flattening is linear, $ \frac{\partial F}{\partial P_1} = I $ (identity), so:
$$
\frac{\partial L}{\partial P_1} = \frac{\partial L}{\partial F}
$$

But now: $ P_1 = \text{maxpool}(A_1) $

Max pooling: only the **maximum element** in each window contributes to the gradient.

So:
$$
\frac{\partial L}{\partial A_1} = \frac{\partial L}{\partial P_1} \cdot \frac{\partial P_1}{\partial A_1}
$$

Where:
- $ \frac{\partial P_1}{\partial A_1} $ is **1 at max location**, 0 elsewhere (in each 2×2 window)

Thus:
$$
\boxed{ \frac{\partial L}{\partial A_1} = \text{gradient only where max occurred} }
$$

This is called **"max-pooling gradient propagation"**.

---

## ✅ Step 5: Backprop Through ReLU

Now: $ A_1 = \text{ReLU}(z_1) $

So:
$$
\frac{\partial A_1}{\partial z_1} =
\begin{cases}
1 & \text{if } z_1 > 0 \\
0 & \text{if } z_1 \leq 0
\end{cases}
$$

So:
$$
\frac{\partial L}{\partial z_1} = \frac{\partial L}{\partial A_1} \cdot \frac{\partial A_1}{\partial z_1}
$$

→ Only non-zero where $ z_1 > 0 $

---

## ✅ Step 6: Backprop Through Convolution

Now: $ z_1 = \text{conv}(X, w_1) + b_1 $

We want:
$$
\frac{\partial L}{\partial w_1} = \frac{\partial L}{\partial z_1} \cdot \frac{\partial z_1}{\partial w_1}
$$

But $ \frac{\partial z_1}{\partial w_1} $ is not simple — it's the **convolution of input `X` with the gradient**.

In fact:
$$
\frac{\partial L}{\partial w_1} = \text{cross-correlation}(X, \frac{\partial L}{\partial z_1})
$$

This is because:
- The output of convolution is $ z_1[i,j] = \sum_{k,l} X[i-k, j-l] \cdot w_1[k,l] $
- So derivative w.r.t. $ w_1 $ is $ X $ convolved with $ \frac{\partial L}{\partial z_1} $

> This is known as **"convolutional backpropagation"** or **"deconvolution"**.

Similarly:
$$
\frac{\partial L}{\partial b_1} = \sum_{i,j} \frac{\partial L}{\partial z_1}[i,j]
$$

→ Sum of gradients over all spatial locations.

---

## 🧠 Final Chain Rule Expressions

The diagram shows two long chains:

### 1. For $ \frac{\partial L}{\partial w_1} $:

$$
\frac{\partial L}{\partial w_1} = \underbrace{(A_2 - y)}_{\partial L / \partial A_2} \times \underbrace{1}_{\partial A_2 / \partial z_2} \times \underbrace{w_2}_{\partial z_2 / \partial F} \times \underbrace{1}_{\partial F / \partial P_1} \times \underbrace{\text{grad\_pool}}_{\partial P_1 / \partial A_1} \times \underbrace{\text{ReLU}'(z_1)}_{\partial A_1 / \partial z_1} \times \underbrace{\text{conv}(X, \cdot)}_{\partial z_1 / \partial w_1}
$$

But simplified:
$$
\frac{\partial L}{\partial w_1} = \text{conv}(X, \delta_1)
$$
where $ \delta_1 = \frac{\partial L}{\partial z_1} $

---

### 2. For $ \frac{\partial L}{\partial b_1} $:

$$
\frac{\partial L}{\partial b_1} = \sum_{i,j} \frac{\partial L}{\partial z_1}[i,j]
$$

→ Just sum over all elements of $ \frac{\partial L}{\partial z_1} $

---

## ✅ Summary: All 6 Derivatives

| Derivative | Formula |
|---------|--------|
| $ \frac{\partial L}{\partial A_2} $ | $ \frac{A_2 - y}{A_2(1 - A_2)} $ |
| $ \frac{\partial L}{\partial z_2} $ | $ A_2 - y $ |
| $ \frac{\partial L}{\partial w_2} $ | $ (A_2 - y) F^T $ |
| $ \frac{\partial L}{\partial b_2} $ | $ A_2 - y $ |
| $ \frac{\partial L}{\partial F} $ | $ (A_2 - y) w_2 $ |
| $ \frac{\partial L}{\partial w_1} $ | $ \text{conv}(X, \frac{\partial L}{\partial z_1}) $ |

And:
- $ \frac{\partial L}{\partial z_1} = \frac{\partial L}{\partial A_1} \cdot \text{ReLU}'(z_1) $
- $ \frac{\partial L}{\partial A_1} $: only non-zero at max pool locations

---

## 💡 Key Takeaways

1. **Backpropagation uses chain rule** across layers.
2. **Each layer has its own derivative rule**:
   - Sigmoid: $ \sigma'(z) = \sigma(z)(1-\sigma(z)) $
   - ReLU: $ \mathbb{1}(z > 0) $
   - Max Pool: gradient goes only to max element
   - Conv: use cross-correlation with input
3. **Gradients flow backward**:
   - Output → FC → Flatten → Max Pool → ReLU → Conv
4. **Final gradients update weights and biases** via:
   $$
   w \leftarrow w - \eta \frac{\partial L}{\partial w}, \quad b \leftarrow b - \eta \frac{\partial L}{\partial b}
   $$

## 🎯 Goal: Learn How to Update the Filter `w₁`

We have:
- An input image `X` (6×6)
- A small filter `w₁` (3×3)
- We slide this filter over the image → get a feature map `z₁` (4×4)

Now, during training, we want to **adjust the filter `w₁`** so that the network makes better predictions.

To do that, we need to know:
> "If I change each number in `w₁`, how does it affect the final loss?"

That’s what **gradient** tells us.

---

## ✅ Step 1: How Each Weight in `w₁` Affects Output

Let’s take one weight in the filter: say `w₁[0,0]` (top-left corner).

When we slide the filter over the image:
- At position (0,0):  
  `z₁[0,0] = X[0,0]*w₁[0,0] + X[0,1]*w₁[0,1] + ... + X[2,2]*w₁[2,2]`
- So `w₁[0,0]` is multiplied by `X[0,0]`

→ So if we increase `w₁[0,0]`, `z₁[0,0]` increases by `X[0,0]` times that change.

So:
> **Effect of `w₁[0,0]` on `z₁[0,0]` = X[0,0]**

Similarly:
- At position (0,1):  
  `z₁[0,1] = X[0,1]*w₁[0,0] + X[0,2]*w₁[0,1] + ...`
- So `w₁[0,0]` is multiplied by `X[0,1]`

→ Effect = `X[0,1]`

And so on...

So `w₁[0,0]` affects:
- `z₁[0,0]` → by `X[0,0]`
- `z₁[0,1]` → by `X[0,1]`
- `z₁[1,0]` → by `X[1,0]`
- `z₁[1,1]` → by `X[1,1]`
- etc.

In general:
> Each time `w₁[0,0]` is used, it multiplies a pixel from `X`.

So total effect = sum over all positions where `w₁[0,0]` was used.

---

## ✅ Step 2: How Change in `z₁` Affects Final Loss

Now, the loss depends on the final prediction `A₂`.

But to get from `z₁` to loss, we go through:
- ReLU → Max Pool → Flatten → FC Layer → Sigmoid → Loss

Let’s call the **gradient of loss w.r.t. `z₁`** as `delta`.

So:
- `delta[0,0]` = how much changing `z₁[0,0]` affects the loss
- `delta[0,1]` = how much changing `z₁[0,1]` affects the loss
- ...
- `delta[3,3]` = for last position

We already computed this `delta` in earlier steps (from backprop through max pool, ReLU, etc.).

So now:
> If I change `z₁[0,0]` a little, the loss changes by `delta[0,0]`

---

## ✅ Step 3: Combine Both Ideas

Now, we want:
> How does changing `w₁[0,0]` affect the loss?

It affects:
- `z₁[0,0]` → which changes loss by `delta[0,0] * X[0,0]`
- `z₁[0,1]` → which changes loss by `delta[0,1] * X[0,1]`
- `z₁[1,0]` → `delta[1,0] * X[1,0]`
- `z₁[1,1]` → `delta[1,1] * X[1,1]`
- and so on...

So total effect:
> Gradient of loss w.r.t. `w₁[0,0]` =  
> `delta[0,0]*X[0,0] + delta[0,1]*X[0,1] + delta[1,0]*X[1,0] + delta[1,1]*X[1,1] + ...`

This is just a **sum of products** — like a dot product.

And guess what? This is exactly what **convolution** does!

---

## ✅ Final Insight: Gradient = Convolution

So:
> To compute the gradient for the filter `w₁`, we do:
>
> **Convolve the input image `X` with the gradient map `delta`**

That is:
$$
\text{gradient for } w₁ = \text{conv}(X, \text{delta})
$$

Or in simple words:
- Slide `delta` over `X`
- At each position, multiply and add
- Result is the gradient for `w₁`

This is shown in the image:  
> `conv(X, dL/dz₁)` gives the gradient for `w₁`

---

## 🧠 Example (Simple Numbers)

Suppose:
- `X` is:
  ```
  1  2  3
  4  5  6
  7  8  9
  ```
- `delta` (gradient from above) is:
  ```
  0.1  0.2
  0.3  0.4
  ```

Now compute gradient for `w₁[0,0]`:
- Overlap top-left of `X` with `delta[0,0]`: multiply `1*0.1`
- Then move: `2*0.2`
- Down: `4*0.3`
- Diagonal: `5*0.4`
- Add them: `1*0.1 + 2*0.2 + 4*0.3 + 5*0.4 = 0.1 + 0.4 + 1.2 + 2.0 = 3.7`

So:
> Gradient for `w₁[0,0]` = 3.7

Repeat for other positions → get full gradient for `w₁`

---

## ✅ Summary (Simple Words)

| Step | What We Do |
|------|-----------|
| 1 | Get `delta = gradient of loss w.r.t. z₁` (from backprop) |
| 2 | Slide `delta` over input `X` |
| 3 | At each position, multiply and add (like convolution) |
| 4 | Result = gradient for filter `w₁` |
| 5 | Update `w₁`: `w₁ = w₁ - learning_rate × gradient` |

---

## 🎉 Final Answer

> **The math behind updating the filter is:**
>
> - Each part of the image contributes to the output using the filter.
> - The loss tells us how wrong we were.
> - We use the **input image** and the **error map (`delta`)** to figure out how to change the filter.
> - This is done by **convolving the input with the error gradient**.
>
> So:  
> **Gradient for filter = conv(input, error_gradient)**

That’s how neural networks learn what patterns to detect!

### 🧩 Given:

- Input image: $ X \in \mathbb{R}^{6\times6} $
- Filter: $ w_1 \in \mathbb{R}^{3\times3} $
- Bias: $ b_1 \in \mathbb{R} $
- Output: $ z_1 = \text{conv}(X, w_1) + b_1 \in \mathbb{R}^{4\times4} $
- $ A_1 = \max(0, z_1) $
- $ P_1 = \text{maxpool}(A_1) $
- $ F = \text{flatten}(P_1) $
- $ z_2 = w_2 F + b_2 $
- $ A_2 = \sigma(z_2) = \frac{1}{1 + e^{-z_2}} $
- $ L = -y \log A_2 - (1-y)\log(1 - A_2) $

---

### 🔻 Backward Pass

1.  
$$
\frac{\partial L}{\partial z_2} = A_2 - y
$$

2.  
$$
\frac{\partial L}{\partial w_2} = \frac{\partial L}{\partial z_2} F^T = (A_2 - y) F^T
$$

3.  
$$
\frac{\partial L}{\partial b_2} = A_2 - y
$$

4.  
$$
\frac{\partial L}{\partial F} = w_2^T (A_2 - y)
$$

5.  
$$
\frac{\partial L}{\partial P_1} = \text{reshape}\left( \frac{\partial L}{\partial F} \right)
$$

6.  
$$
\frac{\partial L}{\partial A_1}[i,j] =
\begin{cases}
\frac{\partial L}{\partial P_1}[k,l] & \text{if } A_1[i,j] \text{ is max in its pooling window} \\
0 & \text{otherwise}
\end{cases}
$$

7.  
$$
\frac{\partial L}{\partial z_1} = \frac{\partial L}{\partial A_1} \cdot \mathbf{1}_{z_1 > 0}
$$

8.  
$$
\frac{\partial L}{\partial w_1}[i,j] = \sum_{p=0}^{3} \sum_{q=0}^{3} X[p+i, q+j] \cdot \frac{\partial L}{\partial z_1}[p,q]
\quad \text{for } i,j = 0,1,2
$$

Or in vector form:
$$
\frac{\partial L}{\partial w_1} = X \star \frac{\partial L}{\partial z_1}
$$
where $ \star $ is cross-correlation.

9.  
$$
\frac{\partial L}{\partial b_1} = \sum_{p=0}^{3} \sum_{q=0}^{3} \frac{\partial L}{\partial z_1}[p,q]
$$

---

### ✅ Final Updates (Gradient Descent)

$$
w_1 \leftarrow w_1 - \eta \frac{\partial L}{\partial w_1}
$$
$$
b_1 \leftarrow b_1 - \eta \frac{\partial L}{\partial b_1}
$$
$$
w_2 \leftarrow w_2 - \eta \frac{\partial L}{\partial w_2}
$$
$$
b_2 \leftarrow b_2 - \eta \frac{\partial L}{\partial b_2}
$$
