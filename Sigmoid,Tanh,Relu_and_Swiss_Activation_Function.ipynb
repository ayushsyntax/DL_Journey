## Why Do We Use Activation Functions in Neural Networks?

Activation functions introduce **non-linearity** into neural networks, enabling them to model complex relationships in data. Without activation functions, neural networks—regardless of how many layers—would reduce to just a linear transformation, no better than simple linear regression. This means:

- The output would always be a **linear function** of the input, limiting the model to learn only linear patterns.
- Non-linear activation functions allow networks to learn richer, more intricate data distributions and solve tasks like image recognition, language understanding, and beyond.[1][2][3][4][5][6]

> "Activation functions are necessary for neural networks because, without them, the output of the model would simply be a linear function of the input... By introducing nonlinearity through activation functions, neural networks are able to model more complex functions within every node, which enables the neural network to learn more effectively."[1]

### What if We Don't Use Activation Functions?

- Networks become equivalent to a **single linear transformation** and cannot solve complex problems.
- Deep architectures lose their advantage, and no matter the depth, the expressive power remains as limited as a single-layer linear model.[4][5]

## Overview: Sigmoid, Tanh, and ReLU Activation Functions

Below is a comparison of each function, including their **need, pros/cons, and graphical illustration**:

| Function | Formula | Output Range | Pros | Cons | Typical Use |
|---|---|---|---|---|---|
| **Sigmoid** | $$ \sigma(x) = \frac{1}{1 + e^{-x}} $$ | (0, 1) | - Output interpretable as probability- Smooth gradient- Used in binary classification output layers | - **Not zero-centered**- **Vanishing gradients** for large |x|- Slower convergence in deep nets- Can cause saturated neurons that don't learn[4][7][8][9][10] | Output layer (binary) |
| **Tanh** | $$ \tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}} $$ | (-1, 1) | - **Zero-centered** output- Stronger gradients in range- Faster convergence- Better for hidden layers than sigmoid | - **Vanishing gradients** for large |x|- Still saturates at extremes- More sensitive to outliers[4][7][11][12] | Hidden layers |
| **ReLU** | $$ f(x) = \max(0, x) $$ | [0, ∞) | - Computationally efficient- Avoids vanishing gradients for positive values- Enables sparse activation- Fastest training | - **Dying ReLU** (neurons stuck at 0)- Not zero-centered- Ignores negative inputs[4][13][5][14][15] | Most hidden layers |

***

## Pros and Cons — In Detail

### Sigmoid
- **Pros:**
  - Good for probabilities.
  - Smooth, continuous output.
- **Cons:**
  - Not zero-centered (can slow training).
  - Outputs saturate for large positive/negative values (vanishing gradient problem).
  - Slow convergence in deep networks.
  - Can be computationally expensive due to exponentials.

### Tanh
- **Pros:**
  - Zero-centered output (balances gradients/better learning).
  - Typically faster convergence than sigmoid.
- **Cons:**
  - Still suffers from vanishing gradients for extreme values.
  - Not suitable for all tasks — still outperformed by ReLU in deep nets.
  - Sensitive to input scaling.

### ReLU
- **Pros:**
  - Fast to compute (simple function).
  - Enables quick training and deep networks.
  - Less likely to suffer from vanishing gradients.
- **Cons:**
  - "Dying ReLU" problem (some neurons permanently output zero).
  - Not zero-centered (all output is nonnegative).
  - Not suitable when negative outputs are needed.

## Graphical Illustration

### Sigmoid Function

The sigmoid curve shows a smooth "S" shape, outputting values between 0 and 1.

$$

### Tanh Function

Similar to sigmoid but output ranges between -1 and 1. Also an "S" shape, but centered at 0.

$$
\tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}
$$



### ReLU Function

ReLU is linear for positive values and 0 for negative values.

$$
f(x) = \max(0, x)
$$

## Conclusion

- **Activation functions** are essential for any neural network's ability to learn non-linear patterns.
- Without them, deep models collapse to simple linear regression — losing the power of deep learning.
- **Choice of activation:**  
  - **Sigmoid** for binary outputs.
  - **Tanh** for hidden layers in shallow networks.
  - **ReLU** is the **standard** for most deep hidden layers due to its efficiency and avoidance of vanishing gradients.[7][13][14][4][1]

**Pro tip:** For very deep nets, use ReLU or its variants to maximize learning power and speed.

#How to Choose the Right Activation Function
To choose an activation function when training a neural network, it is typically a good idea to start with a ReLU-based function, as this function has been successful in a wide range of tasks. However, certain architectures, like RNNs and LSTMs, may require specific activation functions like sigmoid or tanh.

It is recommended to use the ReLU activation function only in the hidden layers of a neural network. In addition, Sigmoid/Logistic and Tanh functions should not be used in the hidden layers of a neural network because they can cause issues with training due to vanishing gradients.

Here are some guidelines for selecting an activation function for the output
layer based on the type of prediction problem:

Regression: Use Linear Activation Function

Binary Classification: Use Sigmoid/Logistic Activation Function

Multi-class Classification: Use Softmax

Multi-label Classification: Use Sigmoid
